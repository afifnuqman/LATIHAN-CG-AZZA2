```python
nama = input("Masukkan nama anda: ")
umur = input("Masukkan umur anda: ")

print(f"Halo {nama}, umur anda ialah {umur} tahun.")
```

```python

panjang = float(input("Masukkan panjang segi empat: "))
lebar = float(input("Masukkan lebar segi empat: "))

luas = panjang * lebar


print(f"Luas segi empat ialah: {luas}")
```

```python

nombor = int(input("Masukkan satu nombor bulat: "))

if nombor % 2 == 0:
    print(f"Nombor {nombor} ialah nombor GENAP.")
else:
    print(f"Nombor {nombor} ialah nombor GANJIL.")
```

```python
markah = float(input("Masukkan markah anda (0-100): "))

if markah >= 40:
    print("Status: LULUS 🎉")
else:
    print("Status: GAGAL ❌")
```
