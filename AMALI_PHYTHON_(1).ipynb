```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
markah = np.array([70, 85, 90, 65, 80])
print(markah)
```

BINA ARRAY MARKAH

```python
purata = np.mean(markah)
print("Purata markah:", purata)
```

KIRA PURATA MARKAH

```python
tertinggi = np.max(markah)
print("Markah tertinggi:", tertinggi)
```

CARI MARKAH TERTINGGI

```python
terendah = np.min(markah)
print("Markah terendah:", terendah)
```

CARI MARKAH TERENDAH

```python
jumlah = np.sum(markah)
print("Jumlah markah:", jumlah)
```

Kira jumlah keseluruhan markah

```python
markah_baru = markah + 5
print(markah_baru)
```

```python
data = {
"Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
"Markah": [70, 85, 90, 65, 80]
}
```

 Sediakan data pelajar

```python
df = pd.DataFrame(data)
print(df)
```

Bina DataFrame

```python
print(df.head())
```

Paparkan data teratas

```python
print(df["Nama"])
```

Paparkan nama pelajar sahaja

```python
print(df["Markah"])
```

Paparkan markah sahaja

```python
purata = df["Markah"].mean()
print("Purata markah:", purata)
```

Kira purata markah menggunakan pandas

```python
pelajar_cemerlang = df[df["Markah"] >= 80]
print(pelajar_cemerlang)
```

Tapis pelajar yang mendapat markah 80 dan ke atas

```python
df["Status"] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")
print(df)
```

Tambah lajur status

```python
def tentukan_gred(markah):
    if markah >= 80:
        return "A"
    elif markah >= 65:
        return "B"
    elif markah >= 50:
        return "C"
    else:
        return "D"

df["Gred"] = df["Markah"].apply(tentukan_gred)
print(df)
```

Tambah lajur gred

```python
df.to_csv("markah_pelajar.csv", index=False)
print("Fail berjaya disimpan.")
```
